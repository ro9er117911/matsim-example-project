# Task
Create helper functions to convert timestamps (in seconds) to `HH:MM:SS` format, transform coordinates from "epsg:4326" (latitude, longitude) to "EPSG:3826", and map numeric transportation mode codes to XML-specific text modes (e.g., 'BUS' and 'RAIL' to 'pt', 'CAR' to 'car', 'WALK' to 'walk').

## 定義輔助函數 (含座標轉換)

### Subtask:
創建輔助函數來將時間戳（秒數）轉換為 HH:MM:SS 格式。同時，定義一個函數將 "epsg:4326" 座標（經緯度）轉換為 "EPSG:3826" 座標，並將數字模式代碼映射到 XML 格式所需的文本模式（例如，將 'BUS' 和 'RAIL' 映射為 'pt'，'CAR' 映射為 'car'，'WALK' 映射為 'walk'）。


**Reasoning**:
I need to define three helper functions: one for converting seconds to HH:MM:SS, another for transforming coordinates from EPSG:4326 to EPSG:3826, and a third for mapping numeric mode codes to XML-specific text modes, as specified in the subtask instructions.



```python
from pyproj import CRS, Transformer
import datetime

# 1. Function to convert seconds to HH:MM:SS format
def convert_seconds_to_hhmmss(seconds):
    """Converts an integer number of seconds into HH:MM:SS format."""
    return str(datetime.timedelta(seconds=int(seconds)))

# 2. Initialize a global Transformer object for coordinate conversion
transformer = Transformer.from_crs(CRS("epsg:4326"), CRS("epsg:3826"), always_xy=True)

# 3. Function to transform coordinates from EPSG:4326 to EPSG:3826
def transform_coordinates_4326_to_3826(longitude, latitude):
    """Transforms latitude and longitude from EPSG:4326 to EPSG:3826.

    Args:
        latitude (float): Latitude in EPSG:4326.
        longitude (float): Longitude in EPSG:4326.

    Returns:
        tuple: (x_3826, y_3826) transformed coordinates in EPSG:3826.
    """
    x_3826, y_3826 = transformer.transform(longitude, latitude)
    return x_3826, y_3826

# 4. Dictionary to map numeric mode codes to XML-specific text modes
mode_code_to_xml_mode = {
    0: 'walk',  # WALK
    1: 'car',   # CAR
    3: 'pt',    # BUS
    4: 'pt'     # RAIL
}

# 5. Function to map numeric mode codes to XML-specific text modes
def map_mode_to_xml(mode_code):
    """Maps a numeric transportation mode code to its XML-specific text mode."""
    return mode_code_to_xml_mode.get(mode_code, 'unknown')

print("Helper functions and mapping for time, coordinates, and modes defined.")
```

## 轉換資料為 XML (含座標轉換)

### Subtask:
迭代 `df_parquet` 中的每一行資料。對於每個代理人，根據其 `modes`、`timestamps` 和 `geometry` 欄位，構建一個包含 `<person>`、`<plan>`、`<activity>` 和 `<leg>` 元素的 XML 片段。在構建 `<activity>` 元素時，應用座標轉換函數將位置從 "epsg:4326" 轉換為 "EPSG:3826"。第一項活動將標記為 'home'，中間活動標記為 'transit_point'，最後一項活動也標記為 'home'。


**Reasoning**:
To convert the `df_parquet` data into the specified XML format, I need to iterate through each row, extract the relevant data, and construct the XML string element by element, applying the previously defined helper functions for timestamp formatting, coordinate transformation, and mode mapping.



```python
xml_data = []

for index, row in df_parquet.iterrows():
    agent_id = row['agent_id']
    modes = row['modes']
    timestamps = row['timestamps']
    geometry = row['geometry']

    # Start XML for current agent
    xml_string = f'<person id="{agent_id}">\n'
    xml_string += '  <plan type="typical">\n'

    # 1. Create the first <activity> element (type='home')
    first_geo = geometry[0]
    # The geometry dictionary has 'x' as latitude and 'y' as longitude
    lon_4326, lat_4326 = first_geo['y'], first_geo['x']
    x_3826, y_3826 = transform_coordinates_4326_to_3826(lon_4326, lat_4326)
    end_time_hhmmss = convert_seconds_to_hhmmss(timestamps[0])
    xml_string += f'    <activity type="home" x="{x_3826:.2f}" y="{y_3826:.2f}" end_time="{end_time_hhmmss}"/>\n'

    # 2. Loop through intermediate legs and activities
    # Loop from 0 up to second to last segment (len(modes) - 1)
    for i in range(len(modes) - 1):
        # Create <leg> element
        xml_mode = map_mode_to_xml(modes[i])
        departure_time_hhmmss = convert_seconds_to_hhmmss(timestamps[i])
        xml_string += f'    <leg mode="{xml_mode}" departure_time="{departure_time_hhmmss}"/>\n'

        # Create intermediate <activity> element (type='transit_point')
        next_geo = geometry[i+1]
        lon_4326, lat_4326 = next_geo['y'], next_geo['x']
        x_3826, y_3826 = transform_coordinates_4326_to_3826(lon_4326, lat_4326)
        # Use the corresponding timestamp for the activity's start and end time
        activity_time_hhmmss = convert_seconds_to_hhmmss(timestamps[i+1])
        xml_string += f'    <activity type="transit_point" x="{x_3826:.2f}" y="{y_3826:.2f}" start_time="{activity_time_hhmmss}" end_time="{activity_time_hhmmss}"/>\n'

    # 3. Create the final <leg> element
    if len(modes) > 0: # Ensure there's at least one leg
        last_mode = modes[-1]
        xml_mode = map_mode_to_xml(last_mode)
        # The departure time for the last leg corresponds to the second to last timestamp
        final_leg_departure_time_hhmmss = convert_seconds_to_hhmmss(timestamps[len(modes)-1])
        xml_string += f'    <leg mode="{xml_mode}" departure_time="{final_leg_departure_time_hhmmss}"/>\n'

    # 4. Create the final <activity> element (type='home')
    final_geo = geometry[-1]
    lon_4326, lat_4326 = final_geo['y'], final_geo['x']
    x_3826, y_3826 = transform_coordinates_4326_to_3826(lon_4326, lat_4326)
    xml_string += f'    <activity type="home" x="{x_3826:.2f}" y="{y_3826:.2f}"/>\n'

    # Close XML tags for current agent
    xml_string += '  </plan>\n'
    xml_string += '</person>'
    xml_data.append(xml_string)

print(f"Generated XML for {len(xml_data)} agents.")
# Optionally, print the first XML string to inspect its structure
# if xml_data:
#     print("\nExample XML for the first agent:")
#     print(xml_data[0])
```

## 組合為完整 XML 文件

### Subtask:
將所有代理人生成的 XML 片段組合在一個根元素（例如 <population>）下，形成一個完整的 XML 文件。


**Reasoning**:
I need to combine all the individual XML agent snippets stored in `xml_data` into a single XML document, encased within a `<population>` root element, as per the instructions.



```python
full_xml_document = """<?xml version="1.0" encoding="UTF-8"?>
<!DOCTYPE population SYSTEM "http://www.matsim.org/files/dtd/population_v6.xsd">
<population name="generated_population">
"""

for agent_xml in xml_data:
    # Indent each agent's XML for readability
    # Split by lines, indent each line, then join back
    indented_agent_xml = "  " + "\n  ".join(agent_xml.split('\n'))
    full_xml_document += indented_agent_xml + "\n"

full_xml_document += "</population>"

print("Full XML document generated. First 500 characters:")
print(full_xml_document[:500])
print("\nLast 200 characters:")
print(full_xml_document[-200:])
```
